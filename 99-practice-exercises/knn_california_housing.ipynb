Short in-class exercise exploring the California housing sample data with scatter matrices and pair plots, then fitting a KNN classifier.

*Note:* KNN *classification* was used here to predict a continuous variable (population), hence the low score. A `KNeighborsRegressor` would be the right tool. Kept as a record of practice.

```python
from sklearn import datasets #  Skelearn Library for machine learning - datasets are example inthis library  so there are no standard codes
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
housing=pd.read_csv('/content/sample_data/california_housing_test.csv')
housing
```

```python
housing['target'] = housing.median_house_value
housing
```

```python
import matplotlib.pyplot as plt
pd.plotting.scatter_matrix(housing, c=housing['target'], figsize=[11, 11], s=150)
plt.show()
```

```python
import seaborn as sb
sb.pairplot(housing, hue='target')
```

```python
x = housing[['population', 'households']]
plt.scatter(x['population'], x['households'], c=housing['target'])
plt.show()
```

```python
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier(n_neighbors=5, metric='minkowski',p=2)
y = housing.population
x = housing[['households']]
knn.fit(x, y)
```

```python
knn.score(x,y)
```
