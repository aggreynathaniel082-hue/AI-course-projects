# Visualising categorical survey results

**Goal:** Turn a table of response counts into readable charts.

**Data:** counts of responses (very negative → very positive) across six themes: consumer adoption, cost and affordability, environmental impact, infrastructure readiness, policy and regulation, technological advancement. A second small table shows education levels.

**Methods:** pandas DataFrames from NumPy arrays, bar, horizontal bar, pie and scatter charts.

**Takeaway:** *Environmental impact* and *infrastructure readiness* drew by far the most responses, and the most negative ones, so they are the themes to dig into first.

```python
import pandas as pd
import numpy as np
```

```python
my_series = pd.Series([10, 20, 30, 40, 50])
```

```python
my_array = np.array([[1,3,5,9],[5,3,9,6]])
my_array
```

```python
my_array = np.array([[239,95,344,80,100],[147,62,204,34,36],[1357,255,974,117,235],[711,259,1125,209,274],[148,36,219,32,31],[122,24,206,47,106]])
my_array
```

```python
df = pd.DataFrame(my_array, index = ['Consumer Adoption','Cost and Affordability','Environmental Impact','Infrastructure Readiness','Policy and Regulation','Technological Advancement']
                  ,columns=['Very Negative','Negative','Neutral','Positive','Very Positive'])
df
```

```python
import matplotlib.pyplot as plt
import numpy as np
plt.plot(df['Very Negative'], df['Very Negative'])
plt.xlabel('Index')
plt.ylabel('Value')
plt.title('Plot of Very Negative and Negative Columns')
plt.show()
```

```python
plt.scatter(df['Very Negative'],df['Negative'])
```

```python
plt.figure(figsize=(12, 8), dpi=100)
df[['Very Negative', 'Negative', 'Neutral', 'Positive', 'Very Positive']]
plt.bar(df.index,df['Very Negative'], label='Very Negative')
plt.bar(df.index,df['Positive'], label='Neutral')
plt.xlabel('')
plt.ylabel("bar height")
plt.title("Bar Chart Example")
```

```python
plt.pie(df['Negative'], labels=df.index,autopct='%0.2f%%')
plt.show()
```

```python
import numpy as np
my_array = np.array([[123, 4.1],[307, 10.2],[454, 15.0]])
my_array
```

```python
import pandas as pd
df = pd.DataFrame(my_array, index = ['Less than high school','High school graduate/GED','Some college'],columns =['N', '%'])
df
```

```python
import matplotlib.pyplot as plt
import numpy as np
plt.plot(df['N'], df['%'])
plt.xlabel('N')
plt.ylabel('%')
plt.title('Plot of N vs Percentage')
plt.show()
```

```python
plt.scatter(df['N'], df['%'])
```

```python
plt.pie(df['N'], labels=df.index,autopct='%0.2f%%')
```

```python
plt.pie(df['N'], labels=df.index)
```

```python

plt.barh(df.index, df['N'])
plt.xlabel('N')
plt.ylabel('Education Level')
plt.title('Count (N) by Education Level')
plt.show()
```
