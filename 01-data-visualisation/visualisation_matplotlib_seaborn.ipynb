**Takeaways**

```python
# prompt: how we can use pandas functions for  the upper and lower extrem values in this data frame for each column

# Assuming 'smartphones' is your DataFrame

# Find the upper and lower extremes (max and min) for each column
for column in smartphones.columns:
  if pd.api.types.is_numeric_dtype(smartphones[column]):  # Check if the column is numeric
    upper_extreme = smartphones[column].max()
    lower_extreme = smartphones[column].min()
    print(f"Column: {column}, Upper Extreme: {upper_extreme}, Lower Extreme: {lower_extreme}")

# You can also find the quantiles (e.g., 25th and 75th percentiles) and use the interquartile range (IQR)
# to identify outliers (values outside 1.5 * IQR from the lower and upper quartiles).

# Example using the IQR for 'Ram' column:
Q1 = smartphones['Ram'].quantile(0.25)
Q3 = smartphones['Ram'].quantile(0.75)
IQR = Q3 - Q1
upper_bound = Q3 + 1.5 * IQR
lower_bound = Q1 - 1.5 * IQR

outliers = smartphones[(smartphones['Ram'] > upper_bound) | (smartphones['Ram'] < lower_bound)]
print("Outliers in Ram:", outliers)
```

```python
# prompt: give me the code from pandas to show outlaiers for the column 'ram'

# Assuming 'smartphones' is your DataFrame

# Calculate the IQR for the 'Ram' column
Q1 = smartphones['Ram'].quantile(0.25)
Q3 = smartphones['Ram'].quantile(0.75)
IQR = Q3 - Q1

# Define the lower and upper bounds for outliers
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Identify outliers based on the bounds
outliers = smartphones[(smartphones['Ram'] < lower_bound) | (smartphones['Ram'] > upper_bound)]

# Print the outliers
print("Outliers in Ram:", outliers)

# You can also visualize the outliers using a box plot as you've already done:
sb.boxplot(y='Ram', data=smartphones)
plt.show()
```
